### **Intro to Logging (Real-World Tracking)**

**so logging can be done using multiple method**
-  1. using python logging module 
-  2. using default log in databricks
-  3. the best method is used to create your own logging table and log the data which is required(widely used in industry)

--------------
**This is the correct approach in real Data Engineering environments**
- ✔ Use Python logging for notebook debugging
- ✔ Use Databricks Job logs for platform issues
- ⭐ BEST: Create your own Delta log table for pipeline logs
- ⭐ BEST: Add monitoring dashboards on top of log table

```python

import logging

#this level means this is the basic level and afte that level will increase as we move down , this python inbuilt module help us print the log details into the terminal 
logging.basicConfig(level=logging.DEBUG)

logging.debug("🛠 Debug — Internal variable set")
logging.info("ℹ️ Info — Starting process")
logging.warning("⚠️ Warning — Using default config")
logging.error("❌ Error — Invalid input received")
logging.critical("🔥 Critical — System failure")
```

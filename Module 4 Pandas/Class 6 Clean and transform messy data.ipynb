```python
#handle the amount colum - Filling the missing value
print(df['amount'].describe())
```

```python
df['amount'].describe()
```

```python
#Fill the missing values with median 
#median will first sort the values and ignore the null part and then find the middle value and replace all the null/nan values with the median value
df['amount'] = df['amount'].fillna(df['amount'].median())
display(df['amount'])
```

```python
display(df)
```

### now we are fixing the date column - parse and clean invalid entries

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')
display(df)
```

### Now some date format were not consistent with the default format that to_datetime use so there are chances that those dates are not being converted to datetime format 
_Now we will solve this problem_

```python
df['date'] = pd.to_datetime(df['date'], errors='coerce')
print("\n\n\nInvalid Dates: " , df['date'].isnull().sum())

#fill this null dates
df['date'] = df['date'].fillna(method= 'ffill')

display(df)
```

```python
df.dtypes
```

### Add some usefull columns for analyts like year and column

```python
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day'] = df['date'].dt.day
display(df)
```

### Final step is always to do a sanity check

```python
print(df.info())
print(df.describe())
print(df.head())
```

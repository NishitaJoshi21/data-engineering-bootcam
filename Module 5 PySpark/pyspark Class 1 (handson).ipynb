# Understand DataFrame in Spark 
### Diference between DataFrame in Pysaprak and Pandas

##### Here u can see that in Spark DF we got datatypes as String and long while in Pandas we get it like object and int64 ..so this is one of major difference

## *Interview Oues* - Difference between spark and Pandas DF

```python
# Create DataFrame in PySpark

data = [('Aman',45000), ('Priya',80000),('Anil',30000),('Sneha', 60000)]
column = ['Name','Salary']

spark_df = spark.createDataFrame(data, column)
display(spark_df)
```

### Create a pandas DF

```python
data = [('Aman',45000), ('Priya',80000),('Anil',30000),('Sneha', 60000)]
column = ['Name','Salary']

import pandas as pd
df = pd.DataFrame(data, columns=column)
display(df)
```

#### *So if I want to create a spark DataFrame using tupe, list , dict than using this method we can explicitly defined our schema that , that DataFrame should follow*

```python
# We can also add explicity schema and ask our DataFrame to follow it 
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

schema = StructType([
    StructField("Name", StringType() , True),
    StructField("Salary", IntegerType(), True)
])

df_type = spark.createDataFrame(data, schema= schema)
df_type.printSchema()
df_type.show()
```

```python
display(df_type)
```

```python
#Build a PipelineStep class

class PipelineStep:
    def __init__(self, step_name):
        # So every PipelineStep object will have its own step_name.
        self.step_name = step_name
    
    def execute(self):
        print(f"Executing Step: {self.step_name}")

# Creating Objects
ingest = PipelineStep("Ingestion")
transform = PipelineStep("Transformation")

# Calling the Method
ingest.execute()
transform.execute()
```

## Encapsulation – Protect Internal Logic

**👉 Focus: hiding data using protected/private variables
❌ Not focused on: hiding method logic
Encapsulation = data protection.

🟩 Real-life example of encapsulation
ATM machine:
- You enter PIN → PIN is private
- You cannot see the internal storage of money
- Machine protects internal details

Data is protected → that's encapsulation.**

```python
#create a class and perform Encapsulation
class DatabaseConnector:
  def __init__(self):
    self._credential = "user:pass@123"   #protected Variable credential
  
  def connect(self):
    print("Connection using credential...")
    return "DB connection established"

db = DatabaseConnector()
print(db.connect())
# print(db.credential)  #it is a protected variable so i cannot access it like this 
```

### #Access Modifiers

```python
#Access Modifiers
class DatabaseConnector:
    def __init__(self):
        self._credential = "user:pass@123"   # Protected
        self.__token = "abcd1234"            # Private

    # Public Method
    def connect(self):
        print("Connection using credential...")
        return "DB connection established"

    # Getter for private variable
    def get_token(self):
        return self.__token

db = DatabaseConnector()
print(db.connect())


print(db._credential)   # Allowed but not recommended
print(db.get_token())   # Acess private variable
# print(db.__token)  # Error


```

## Inheritance – Reuse Code with Variants

```python
class reader:
  def read(self):
    return "Reading from the base file.."

class CSVReader(reader):
  def read(self):
    return "Reading from the CSV file.."
  
class JSONReader(reader):
  def read(self):
    return "Reading from JSON file.."

print(CSVReader().read())
print(JSONReader().read())


# polymorphism 
# def run_reader(reader):
#     print(reader.read())

# run_reader(CSVReader())
# run_reader(JSONReader())
```

## # Polymorphism – Same Interface, Different Behavior

```python

def run_reader(reader):
    print(reader.read())

run_reader(CSVReader())
run_reader(JSONReader())
```

## Abstraction – Enforce Structure Across 

**👉 Focus: What the object does
❌ Not focused on: How it does it
You hide internal logic and only show the essential functionality.

🟩 Real-life example of abstraction
When you drive a car:
You press accelerator pedal
Car speeds up 🚗💨

But do you know how the engine:
- mixes fuel
- compresses air
- ignites combustion
- turns the crankshaft?
No

Those details are hidden → that's abstraction.**

```python
from abc import ABC, abstractmethod

class Ingestor(ABC):
    @abstractmethod
    def read(self):
        pass

class CSVIngestor(Ingestor):
    def read(self):
        return "Reading CSV..."

class APIIngestor(Ingestor):
    def read(self):
        return "Calling API..."

def run(ingestor):
    print(ingestor.read())

run(CSVIngestor())
run(APIIngestor())
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):
    
    @abstractmethod
    def pay(self):
        pass

class UPI(Payment):
    def pay(self):
        return "Paying using UPI"

class Card(Payment):
    def pay(self):
        return "Paying using Credit Card"

def make_payment(payment):
    print(payment.pay())

make_payment(UPI())
make_payment(Card())
```

## Real Data Pipeline: OOP Version

```python


class FileIngestor:
  def __init__(self, path):
    self.path = path

  def read(self):
    print(f"Reading data from {self.path}")
    return f"Raw data from {self.path}"
  
class Cleaner:
  def clean(self, data):
    print("Cleaning data...")      
    return f"Cleaned version of: {data}"
  
class Writer:
  def write(self, data):
    print(f"Writing data: {data}")
    return "Write success"    
  
# Execute pipeline
reader = FileIngestor("data.csv")
raw_data = reader.read()

cleaned = Cleaner().clean(raw_data)
Writer().write(cleaned)
```

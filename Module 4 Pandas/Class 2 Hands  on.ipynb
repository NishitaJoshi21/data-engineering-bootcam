## **Interview Quetsions**
**_Ouestion: What is the difference between series and list? **_

Answer: 

-     A list is just a collection of values
-     A series has index + values and supports vectorized operations, missing values and metadata

```python
#Creating a custom index in series 
employee = pd.Series([50000, 60000, 700000], index= ['John', 'Jane', 'Jack'])
print(employee)
```

```python
print(employee['Jane'])   # this is how we can access data based on indexes
```

## Series with mixed types

```python
details = pd.Series(["Nishita", 26000, 'Software Engineer', 'Unnited States'])
print(details)
```

## Apply Math on Series


```python
salaries = pd.Series([50000, 60000, 70000])
updated_salaries = salaries * 1.10  #10% increase
print(updated_salaries) 
```

## Handling the missing values in Series

```python
data = pd.Series([100, None, 300])
print("Original data \n", data)
```

```python
#we want to print which index have null value 
print("is null \n", data.isnull())
```

_**we can fill the empty value with some other value**_

```python
print("Fill empty value \n", data.fillna(400));
```

##  DATAFRAMES in _PNDAS_

```python
#creating a dataframe
data = {"name":["Nishita", "Priya", "Priyanka"],
        "age": [26, 25, 24],
        "salary": [50000, 60000, 70000],
        "designation": ["Data Scientist", "Data Engineer", "Data Analyst"]}
print(data)
```

```python
df= pd.DataFrame(data)
display(df)
print(df)
```

## Access columns and rows

```python
#column level accesss

df['salary']
```

```python
display(df['salary'])
```

```python
#row level accesss
#using the lable
print(df.loc[0])
```

```python
#using the index
print(df.iloc[1])
```

## Adding one more column

```python
df['Bonus'] = df['salary'] * 0.10
display(df)
```

```python
# Fill nulls with default values

df['department'] = df['department'].fillna('Unknown')
df['salary'] = df['salary'].fillna(df['salary'].median())

display(df)
```

### convert and clean date column

```python
df['join_date'] = pd.to_datetime(df['join_date'], errors='coerce') #convert invalid date to NaT
df['join_date'] = df['join_date'].fillna(method='ffill')  #fill Nat wil forward fill
display(df)
```

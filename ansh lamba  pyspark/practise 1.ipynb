# Read Json
1️⃣ header=True → ❌ NOT USED FOR JSON
header is only for CSV
JSON has keys, not headers
👉 Spark ignores this option silently

2️⃣ inferSchema=True → ❌ Redundant for JSON
Spark always infers schema for JSON
There is no schema-less JSON read
So your code works, but two options are useless

```python
file = spark.read.json("/Volumes/data-bootcamp/default/all_csv/drivers.json",
                       inferSchema=True,
                       header=True,
                       multiLine=False)
file.display()
```

# Schema defination
we can do schema using 2 ways 
1. ddl schema
2. structtype schema (mostly used)

```python
ddl_schema = '''
                    Item_Identifier STRING,
                    Item_Weight STRING,
                    Item_Fat_Content STRING, 
                    Item_Visibility DOUBLE,
                    Item_Type STRING,
                    Item_MRP DOUBLE,
                    Outlet_Identifier STRING,
                    Outlet_Establishment_Year INT,
                    Outlet_Size STRING,
                    Outlet_Location_Type STRING, 
                    Outlet_Type STRING,
                    Item_Outlet_Sales DOUBLE 

'''

df = spark.read.csv("/Volumes/data-bootcamp/default/all_csv/BigMart Sales.csv",schema=ddl_schema, header = True)
df.display()
```

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *

struct_schema = StructType([
    StructField("Item_Identifier",StringType(), True),
    StructField('Item_Weight',StringType(),True),
    StructField('Item_Fat_Content',StringType(),True), 
    StructField('Item_Visibility',StringType(),True),  #double to string
    StructField('Item_MRP',StringType(),True), 
    StructField('Outlet_Identifier',StringType(),True), 
    StructField('Outlet_Establishment_Year',StringType(),True), 
    StructField('Outlet_Size',StringType(),True), 
    StructField('Outlet_Location_Type',StringType(),True), 
    StructField('Outlet_Type',StringType(),True), 
    StructField('Item_Outlet_Sales',StringType(),True),
    StructField('Item_Type',StringType(),True)   #double to string
])

df = spark.read.csv("/Volumes/data-bootcamp/default/all_csv/BigMart Sales.csv",schema = struct_schema, header = True)
df.display()
```

# Filter/where

1. where the data with fat_content = regular
2. slice the data with item type = soft_drink and weight<10
3. fetch the data with teir in (Teir 1 , Teir 2) and outlet size is null



```python
df.filter(col("Item_Fat_Content")=="Regular").display()
```

```python
df.filter((col("Item_MRP")== 'Soft Drinks') & (col("Item_Weight") <10)).display()
```

```python
df = df.withColumn('Item_Weight', df["Item_Weight"].cast('double'))
```

```python
df.filter((col("Item_MRP")== 'Soft Drinks') & (col("Item_Weight") <10)).display()
```

```python
df.filter((col("Outlet_Location_Type").isin("Tier 1","Tier 2")) & (col("Outlet_Size").isNull())).display()
```

# Create or own SparkSession and App

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('Spark DataFrame').getOrCreate()
print("Session Created", spark.version)
```

```python
data = [("Nishita","IT",5000),
      ("Nisha","HR",2000),
      ("Aman","Consultancy",25000),
      ("Ram","Finance",10000),
      ("Joe","IT",25000),
      ("Joseph","Accounts",50000),
      ("Anna","Data Engineer",75000)
      ]

column= ['Name','Department','Salary']
df = spark.createDataFrame(data,column)

display(df)
```

```python
df.printSchema()
```

## Now we will use Select Function

```python
display(df.select("Name"))
```

```python
# Now select Multiple columns
display(df.select("Name","Salary"))
```

# Filter
### filter(): It is a function which filters the columns/row based on SQL expression or condition.



```python
#filter based on Salary >10000
display(df.filter(df['Salary'] > 10000))
```

```python
#Method 2
display(df.filter(df.Salary>10000))
```

### Apply multiple filter

```python
display(df.filter((df['Salary']>1000) & (df['Department']=='Finance')))
```

# Ques - Which is better Pandas or Spark

## Create a Derived column (col based on another col)

```python
from pyspark.sql.functions import col

# Create col Bonus with 10% of Salary
df = df.withColumn("Bonus", col('Salary') * 0.1)
display(df)
```

## Group By and Aggregation

### Q- Avg of salary by department

```python
display(df.groupBy('Department').avg('Salary'))
```

### Apply multiple aggregation

```python
from pyspark.sql.functions import sum, avg,  min, max

display(df.groupBy("Department").agg(
    sum("Salary").alias("Sum_salary"),
    avg("Salary").alias("Avg_salary"),
    min("Salary").alias("Min_salary"),
    max("Salary").alias("Max_salary")
))
```

# JOINS IN PYSPARK

```python
dept_data = {("IT",105), ("HR",106), ("Finance",107 )}
dept_col = ["Department", "Dept_id"]

def_dept = spark.createDataFrame(dept_data, dept_col)
display(def_dept)
```

```python
# Join Operation
df_join = df.join(def_dept, on="Department",how="inner")
display(df_join)
```

### Student Practice Assignment
📂 *Build a DataFrame with the following schema: Name, Department, Salary, Location*

```python
# Sample data
data = [
    ("Anurag", "IT", 70000, "Bangalore"),
    ("Priya", "HR", 55000, "Mumbai"),
    ("Ravi", "Finance", 65000, "Delhi"),
    ("Sneha", "IT", 60000, "Hyderabad")
]

columns = ["Name", "Department", "Salary", "Location"]

df_task = spark.createDataFrame(data, columns)

# Task 1: Filter IT employees with salary > 60K
df_task.filter((col("Department") == "IT") & (col("Salary") > 60000)).show()

# Task 2: Add column "Hike_Amount" as 15% of salary
df_task = df_task.withColumn("Hike_Amount", col("Salary") * 0.15)
df_task.show()

# Task 3: Group by Department and show average salary
df_task.groupBy("Department").avg("Salary").show()
```

```python
df.select('name','salary').show()
```

```python
#Rename column dept to department.
df.withColumnRenamed('dept', 'department').display()
```

```python
#Add a new column bonus = salary * 0.10.
df.withColumn('bonus',df.salary*0.10).display()
```

```python
#Change data type of salary to Integer.
df.withColumn('salary',df.salary.cast('int')).display()
```

```python
# Drop the dept column.
df.drop('dept').display()
```

```python
# Filter employees with salary > 50,000.
df.filter(df.salary>50000).display()
```

```python
#Filter employees from IT department.
df.filter(df.dept=='IT').display()
```

```python
# Filter employees whose name starts with "A".
df.filter(df.name .startswith('r')).display()
```

```python
# Filter employees with salary between 30,000 and 70,000.
df.filter(df.salary.between(30000,70000)).display()
```

```python
# Filter rows where dept is NULL.
df.filter(df.dept.isNull()).display()
```

```python
# Replace NULL salary with 0.
df.display()
```
